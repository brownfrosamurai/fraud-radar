Fraud is rare, so **accuracy is the wrong headline metric**. A model that always predicts legit can be >99% accurate while catching zero fraud.

**PR-AUC** (average precision) measures ranking quality on the positive class and stays honest under class imbalance. Precision, recall, and F1 are reported at the BLOCK operating point (`score >= 0.9`), not as a substitute for PR-AUC.

This notebook does not report accuracy.
